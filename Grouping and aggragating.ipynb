# Group by and Aggregating

```python
import pandas as pd
df = pd.read_csv(r"C:\Nazim Khan\Data Analytics\Python using jupyter Notebook\Pandas_main\Flavors.csv")
```

```python
df
```

```python
df.groupby('Base Flavor').mean(numeric_only=True)
```

```python
df.groupby('Base Flavor')[['Flavor','Liked']].count()
```

```python
df.groupby('Base Flavor')[['Flavor Rating','Texture Rating','Total Rating']].sum()
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating':['min','max','sum','count'],'Texture Rating':['min','max','sum','count'] })
```

```python
df.groupby('Flavor').describe()
```
